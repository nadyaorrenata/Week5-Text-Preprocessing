```python
import re
import requests
import string
from bs4 import BeautifulSoup
from collections import Counter

import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer, RegexpStemmer
```

```python
#MENGAMBIL SEMUA TEKS

webrequest = requests.get("http://home.wezm.net")
soup = BeautifulSoup(webrequest.text, "html.parser")

paragraphs = soup.findAll('p')
raw_text = " ".join(p.text for p in paragraphs)
print(raw_text)
```

```python
#LOWER CASING

text_lower = str.lower(raw_text)
print(text_lower)
```

```python
#REMOVE PUNCTUATIONS, NUMBERS, and SYMBOLS

punc = string.punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans(' ', ' ', punc))

hasil_rp = remove_punctuation(text_lower)
hasil_rp = re.sub(r"[^a-z\s]", " ", hasil_rp)       # hapus angka & simbol seperti ¹
hasil_rp = re.sub(r"\s+", " ", hasil_rp).strip()    # rapikan spasi & enter berlebih
print(hasil_rp)
```

```python
#REMOVE STOPWORDS

sw = set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join([word for word in str(text).split()
                     if word not in sw and len(word) > 1])

hasil_st = remove_stopwords(hasil_rp)
print(hasil_st)
```

```python
#STEMMING

rs = RegexpStemmer('ing$|s$|ed$', min = 4)

def stemming(text):
    return " ".join([rs.stem(word) for word in text.split()])
hasil_stem = stemming(hasil_st)
print(hasil_stem)
```

```python
#LEMMATIZATION

lemmatizer = WordNetLemmatizer()

def lemmatization(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])

hasil_lemma = lemmatization(hasil_st)
print(hasil_lemma)
```

```python
#TAGGING
nltk.download('averaged_perceptron_tagger_eng')

tags = nltk.pos_tag(hasil_st.split())
print(tags)
```

```python
#CHUNKING
grammar = "NP: {<DT>?<JJ>*<NN.*>+}"
chunker = nltk.RegexpParser(grammar)

tags = nltk.pos_tag(hasil_rp.split())
tree = chunker.parse(tags)

for subtree in tree.subtrees(filter=lambda t: t.label() == 'NP'):
    print(" ".join(word for word, tag in subtree.leaves()))
```
